```python
import numpy as np
from statsmodels.stats.proportion import proportions_ztest

n = 400
full_rate = 87.52 / 100
min_rate = 84.48 / 100
full_valid = round(full_rate * n)
min_valid = round(min_rate * n)

print(f'Full rules:    {full_valid}/400 valid ({full_rate*100:.2f}%)')
print(f'Minimal rules: {min_valid}/400 valid ({min_rate*100:.2f}%)')

count = np.array([full_valid, min_valid])
nobs = np.array([n, n])
z, p = proportions_ztest(count, nobs)
print(f'\nTwo-proportion z-test:')
print(f'  z = {z:.3f}, p = {p:.4f}')
print(f'  Significant at alpha=0.05: {p < 0.05}')
print(f'  Significant at alpha=0.10: {p < 0.10}')

diff = full_rate - min_rate
se = np.sqrt(full_rate*(1-full_rate)/n + min_rate*(1-min_rate)/n)
ci_low = diff - 1.96*se
ci_high = diff + 1.96*se
print(f'\nDifference: {diff*100:.2f} pp')
print(f'95% CI: [{ci_low*100:.2f}, {ci_high*100:.2f}] pp')

# Also run for all other models for completeness
print('\n--- All models ---')
models = {
    'GPT-OSS-120B':  (96.76, 96.74),
    'Qwen3.6-27B':   (94.24, 94.76),
    'Qwen3.5-122B':  (87.52, 84.48),
    'Llama3.3-70B':  (14.72, 14.24),
}
for model, (fr, mr) in models.items():
    fv = round(fr/100 * n)
    mv = round(mr/100 * n)
    z, p = proportions_ztest(np.array([fv, mv]), np.array([n, n]))
    print(f'{model}: {fr:.1f}% vs {mr:.1f}%  |  z={z:.3f}, p={p:.4f}  |  sig@0.05={p<0.05}')
```
